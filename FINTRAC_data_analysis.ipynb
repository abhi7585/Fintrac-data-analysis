```python
# Update each row in object type columns by splitting at '_' and keeping the first part (English part)
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].astype(str).str.split('_').str[0]

display(df.head()) # Display the first few rows of the DataFrame with updated data
```

```python
# Convert 'YearMonthReportReceived' column to string type (already done in previous cleaning step, can be removed if confident in order)
df['YearMonthReportReceived'] = df['YearMonthReportReceived'].astype(str)
```

```python
# Remove rows where 'PostalCode' is an empty string and create a copy of the filtered DataFrame
df = df[df['PostalCode'] != ''].copy()
display(df.head()) # Display the first few rows of the DataFrame after removing rows with blank postal codes
```

```python
# Separate Year and Month from 'YearMonthReportReceived' column
df['YearReportReceived'] = df['YearMonthReportReceived'].astype(str).str[:4] # Extract the first 4 characters as Year
df['MonthReportReceived'] = df['YearMonthReportReceived'].astype(str).str[4:] # Extract characters from the 5th position onwards as Month
display(df.head()) # Display the first few rows with the new 'YearReportReceived' and 'MonthReportReceived' columns
```

```python
# Convert 'YearMonthReportReceived' to datetime objects for time series analysis
df['YearMonthReportReceived'] = pd.to_datetime(df['YearMonthReportReceived'], format='%Y%m')

# Group by year and month and sum the number of reports to get the trend over time
reports_over_time = df.groupby('YearMonthReportReceived')['NumberOfReports'].sum().reset_index()

# Create an interactive line plot using Plotly Express to visualize the trend
fig = px.line(reports_over_time, x='YearMonthReportReceived', y='NumberOfReports', title='Trend of Number of Reports Over Time')
fig.update_layout(xaxis_title='Date', yaxis_title='Number of Reports')
fig.update_xaxes(dtick="M12", tickformat="%Y") # Set x-axis ticks to show each year
fig.show()
```

```python
# Calculate monthly percentage change in the number of reports
reports_over_time['PercentageChange'] = reports_over_time['NumberOfReports'].pct_change() * 100
reports_over_time['Year'] = reports_over_time['YearMonthReportReceived'].dt.year # Extract Year

# Group by year and sum the number of reports to get yearly totals
yearly_reports = reports_over_time.groupby('Year')['NumberOfReports'].sum().reset_index()
# Calculate yearly percentage change in the number of reports
yearly_reports['YearlyPercentageChange'] = yearly_reports['NumberOfReports'].pct_change() * 100

print("Monthly Percentage Change in Number of Reports:")
display(reports_over_time[['YearMonthReportReceived', 'NumberOfReports', 'PercentageChange']]) # Display monthly changes

print("\nYearly Percentage Change in Number of Reports:")
display(yearly_reports) # Display yearly changes
```

**Trend of Reports Over Time:**

To quantify the observed trend, we calculated the percentage change in the number of reports:

*   **Monthly Percentage Change:** We can see month-over-month fluctuations in the number of reports.
*   **Yearly Percentage Change:** The yearly percentage change shows the overall growth or decline in the number of reports each year. From the data, there's a significant yearly increase in reports, particularly between 2011 and 2012 (45.86%). While there are fluctuations, the trend generally indicates an increase in the number of reports over the years, with a substantial increase of 20.44% between 2022 and 2023.

```python
# Function to map the first letter of a postal code to a province/territory
def get_province(postal_code):
    if pd.isna(postal_code) or postal_code == '':
        return 'Unknown'
    first_letter = str(postal_code)[0].upper()
    if first_letter == 'T':
        return 'Alberta'
    elif first_letter == 'V':
        return 'British Columbia'
    elif first_letter == 'R':
        return 'Manitoba'
    elif first_letter == 'E':
        return 'New Brunswick'
    elif first_letter == 'A':
        return 'Newfoundland and Labrador'
    elif first_letter == 'B':
        return 'Nova Scotia'
    elif first_letter in ['K', 'L', 'M', 'N', 'P']:
        return 'Ontario'
    elif first_letter == 'C':
        return 'Prince Edward Island'
    elif first_letter in ['G', 'H', 'J']: # K is also listed for Ontario and Quebec, need clarification or handle potential overlap
         return 'Quebec'
    elif first_letter == 'S':
        return 'Saskatchewan'
    elif first_letter == 'X':
        return 'Territories (NT/NU)' # Cannot distinguish between NT and NU with only the first letter
    elif first_letter == 'Y':
        return 'Yukon'
    else:
        return 'Unknown'

df['Province_Territory'] = df['PostalCode'].apply(get_province) # Apply the function to create a new 'Province_Territory' column
display(df.head()) # Display the first few rows with the new column
```

```python
# Group by Province_Territory and sum the number of reports
reports_by_province = df.groupby('Province_Territory')['NumberOfReports'].sum().reset_index()
reports_by_province = reports_by_province.sort_values(by='NumberOfReports', ascending=False) # Sort by number of reports

print("Total Number of Reports by Province/Territory:")
display(reports_by_province) # Display the total reports by province/territory
```

```python
# Create a bar chart to visualize the total number of reports by province/territory
fig = px.bar(reports_by_province, x='Province_Territory', y='NumberOfReports',
             title='Total Number of Reports by Province/Territory',
             labels={'Province_Territory': 'Province/Territory', 'NumberOfReports': 'Number of Reports'},
             text=reports_by_province['NumberOfReports'].apply(lambda x: f'{x/1000000:.1f}M')) # Format text labels in millions on top of bars
fig.update_layout(xaxis_title='Province/Territory', yaxis_title='Number of Reports', template='plotly') # Set layout and theme
fig.show() # Display the chart
```

```python
# Group by year and province/territory and sum the number of reports for yearly distribution
reports_yearly_province = df.groupby(['YearReportReceived', 'Province_Territory'])['NumberOfReports'].sum().reset_index()

# Create a stacked bar chart to visualize the yearly distribution of reports by province/territory
fig = px.bar(reports_yearly_province, x='YearReportReceived', y='NumberOfReports', color='Province_Territory',
             title='Yearly Distribution of Reports by Province/Territory',
             labels={'YearReportReceived': 'Year', 'NumberOfReports': 'Number of Reports', 'Province_Territory': 'Province/Territory'})
fig.update_layout(xaxis_title='Year', yaxis_title='Number of Reports')
fig.show() # Display the chart
```

# FINTRAC Data Analysis

This notebook provides an analysis of FINTRAC (Financial Transactions and Reports Analysis Centre of Canada) data. The analysis focuses on understanding trends in the number of reports submitted over time, the geographical distribution of these reports across Canada, and the breakdown of report types within different activity sectors.

## Notebook Structure and Analysis Steps

1.  **Data Loading and Initial Inspection**: The analysis begins by loading the data from an Excel file (`fintrac-canafe_data-donnees.xlsx`) into a pandas DataFrame. Initial steps include previewing the data, checking data types, and identifying duplicate rows.
2.  **Data Cleaning and Preparation**: This section focuses on cleaning the data for analysis. This involves:
    *   Cleaning column names to retain only the English portion.
    *   Cleaning data within columns (object types) to remove French text.
    *   Handling missing or blank postal codes by removing relevant rows.
    *   Extracting Year and Month into separate columns from the `YearMonthReportReceived` column.
    *   Cleaning the 'ActivitySector' column.
3.  **Trend Analysis**: This part of the analysis examines how the number of reports has changed over time.
    *   The total number of reports is aggregated by year and month.
    *   An interactive line plot visualizes the trend of reports over time.
    *   Monthly and yearly percentage changes in report numbers are calculated and displayed to quantify the trend.
4.  **Geographical Distribution Analysis**: This section explores where the reports are originating from within Canada.
    *   The total number of reports is aggregated by Postal Code to identify areas with high reporting volumes.
    *   A new column `Province_Territory` is created based on the first letter of the Postal Code.
    *   The total number of reports is aggregated and visualized by `Province_Territory` using a bar chart.
    *   The top 3 contributing provinces/territories by year are identified and visualized.
5.  **Yearly Distribution by Province and Report Type**: A detailed breakdown of report types within each province/territory over the years is performed and visualized using faceted bar charts for all report types, and then specifically for STR, LCTR, and EFT reports.
6.  **Fiscal Year Trend**: The total number of reports is analyzed and visualized by Fiscal Year.
7.  **Cross-analysis by Activity Sector and Report Type**: The analysis attempts to explore the distribution of report types within different activity sectors, noting the dominance of the 'Banks' sector.

## Data Sources:

*   **FINTRAC Data on Financial Transactions:**
    *   **Dataset:** Financial transaction report counts by postal code and activity sector
    *   **Details:** This dataset provides monthly counts of financial transaction reports submitted to FINTRAC, categorized by activity sector, report type, and reporting entity location.
    *   **Usage:** Use this data to analyze the volume and distribution of various financial transactions across different sectors and regions.
*   **FINTRAC Publications and Reports:**
    *   **Source:** FINTRAC Publications
    *   **Details:** FINTRAC publishes strategic intelligence reports, operational alerts, and sectoral advisories that can provide insights into emerging trends and typologies in money laundering activities.
    *   **Usage:** Incorporate findings from these publications to contextualize your dashboard and highlight areas of concern.

## Key Observations and Findings

📊 **Here’s what the data reveals:**

🔼 1.35+ billion reports were filed between 2011 and 2023, marking a ~70% increase in total reporting volume. The biggest jumps occurred in 2011–2012 (+45.9%) and 2022–2023 (+20.4%) — both tied to major regulatory and operational shifts.

💳 **By Report Type:**

EFT (Electronic Funds Transfers): ~805M reports (≈59%) — dominant and steadily rising.

LCTR (Large Cash Transactions): ~396M reports (≈29%) — declining after 2020 due to reduced cash activity.

STR (Suspicious Transactions): ~151M reports (≈11%) — doubled since 2017, highlighting enhanced AML vigilance.

CDR (Casino Disbursements): ~2M reports — rebounded after pandemic slowdowns.

🏦 **By Sector:**

Banks dominate with over 800M EFT and 396M LCTR filings — driving nearly 90% of total reports.

Money Services Businesses (MSBs) show strong STR concentration (~3M), marking a key AML focus area.

Credit Unions remain consistent contributors (~38M total).

Casinos, Real Estate, and Precious Metals sectors report smaller volumes but carry higher inherent risk.

🌎 **Geographically:**

Ontario leads (~40% of all reports), followed by Quebec (~17%) and British Columbia (~15%).

Activity is heavily concentrated in Toronto, Montreal, and Vancouver — Canada’s major financial hubs.

📅 **Fiscal Trend:**

Total annual reports rose from ~68M in FY2011–12 to ~180M in FY2022–23.

EFTs and STRs are driving this upward curve, underscoring a clear shift toward digital vigilance and risk-based reporting.

In short:

Canada’s AML reporting landscape is maturing rapidly — from cash-heavy monitoring to sophisticated, technology-driven oversight.

The continued rise in STRs and EFTs highlights growing vigilance and stronger compliance systems across the industry.

This analysis provides insights into the reporting patterns captured by FINTRAC data, highlighting trends over time, key geographical areas, and the distribution of different report types.
